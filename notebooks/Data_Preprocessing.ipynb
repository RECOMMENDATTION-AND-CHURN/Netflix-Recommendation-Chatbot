```python
movies = movies[
[
'id',
'title',
'overview',
'tagline',
'genres',
'keywords',
'original_language',
'popularity',
'runtime',
'release_date',
'vote_average',
'cast',
'crew'
]
]
```

```python
movies.dropna(subset=['tagline'], inplace=True)
```

```python
movies.dropna(subset=['crew'], inplace=True)
```

```python
movies.isnull().sum()
```

```python
def extract_names(text):
    names = []

    for item in ast.literal_eval(text):
        names.append(item['name'])

    return ", ".join(names)

movies['genres'] = movies['genres'].apply(extract_names)
```

```python
movies.head()
```

```python
movies['keywords'] = movies['keywords'].apply(extract_names)
```

```python
def extract_cast(text):
    cast_names = []
    if not isinstance(text, str) or pd.isna(text) or not text.strip():
        return ""
    try:
        list_of_dicts = ast.literal_eval(text)
        for item in list_of_dicts[:3]:
            if isinstance(item, dict) and 'name' in item:
                cast_names.append(item['name'])
    except (ValueError, SyntaxError):
        return ""
    return ", ".join(cast_names)

# Ensure 'movies' DataFrame has the 'cast' column by re-merging and selecting if necessary
if 'cast' not in movies.columns:
    print("Warning: 'cast' column not found. Re-merging and re-selecting columns to reconstruct 'movies'.")
    # Re-merge `movies` and `credits`
    movies = movies.merge(credits, on="title")

    # Re-select necessary columns including 'cast'
    movies = movies[[
        'id',
        'title',
        'overview',
        'tagline',
        'genres',
        'keywords',
        'original_language',
        'popularity',
        'runtime',
        'release_date',
        'vote_average',
        'cast',
        'crew'
    ]]
    # Re-apply dropna operations that happened previously
    movies.dropna(subset=['tagline'], inplace=True)
    movies.dropna(subset=['crew'], inplace=True)

movies['cast'] = movies['cast'].apply(extract_cast)
```

```python
def extract_crew(text):
    if not isinstance(text, str) or pd.isna(text) or not text.strip():
        return ""

    important_people = []

    try:
        crew_list = ast.literal_eval(text)

        for person in crew_list:
            if person.get("job") in ["Director", "Original Music Composer"]:
                important_people.append(person.get("name"))

        return ", ".join(important_people)

    except (ValueError, SyntaxError):
        return ""

movies["crew"] = movies["crew"].apply(extract_crew)
```
